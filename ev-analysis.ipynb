```python
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path
```

```python
# start with the notebook's working directory
project_path = Path.cwd()
print(f"Current working directory: {project_path}")

# define the path to the data folder
data_folder = project_path / "Data"

# define the states to include
states = ["NSWACT", "NT", "QLD", "SA", "TAS", "VIC", "WA"]

```

```python
# find all FLEET csv files across states 

csv_files = []

for state in states:
    state_folder = data_folder / state
    csv_files.extend(state_folder.glob("FLEET*.csv"))

csv_files = sorted(csv_files)

for file in csv_files:
    print(file)
```

```python
# add STATE column to each dataframe and combine
dfs = []

for file in csv_files:
    df = pd.read_csv(file, dtype={"POSTCODE": str}
    )

    # identify STATE from folder name 
    state = file.parent.name

    # NSWACT already contains NSW and ACT labels
    # other folders use their folder names as STATE labels

    # add STATE column if missing
    if state != "NSWACT":
        df["STATE"] = state

    # append dataframe to the list 
    dfs.append(df)

# combine all files by matching column names 
ev_df = pd.concat(dfs, ignore_index=True)
```

```python
# check the combined dataframe shape
print("Rows and columns:", ev_df.shape)

# check columns
print(ev_df.columns.tolist())

# check coverage
print("Unique states:", ev_df["STATE"].unique())
```
